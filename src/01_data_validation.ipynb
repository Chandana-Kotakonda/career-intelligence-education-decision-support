```python
from pathlib import Path
import pandas as pd

# Project root
PROJECT_ROOT = Path.cwd().parent

# Dataset paths
ONET_DIR = PROJECT_ROOT / "data" / "external" / "onet"
AISHE_DIR = PROJECT_ROOT / "data" / "external" / "aishe"
JOB_DIR = PROJECT_ROOT / "data" / "raw" / "archive (3)"

print("Project Root:", PROJECT_ROOT)
print("O*NET Folder:", ONET_DIR)
print("AISHE Folder:", AISHE_DIR)
print("Job Data Folder:", JOB_DIR)
```

```python
# O*NET files we expect
onet_files = [
    "occupation_data.csv",
    "essential_skills.csv",
    "transferable_skills.csv",
    "software_skills.csv",
    "education.csv",
    "training_and_experience.csv",
    "career_interest_types.csv",
    "related_occupations.csv",
    "job_titles.csv"
]

print("O*NET DATASET CHECK")
print("=" * 60)

for file_name in onet_files:
    file_path = ONET_DIR / file_name
    
    if file_path.exists():
        print(f"✅ {file_name}")
    else:
        print(f"❌ {file_name} — MISSING")
```

```python
# O*NET data quality check

for file_name in onet_files:
    file_path = ONET_DIR / file_name
    
    df = pd.read_csv(file_path, low_memory=False)
    
    print("\n" + "=" * 70)
    print(file_name)
    print("=" * 70)
    
    print(f"Rows              : {len(df):,}")
    print(f"Columns           : {len(df.columns)}")
    print(f"Missing values    : {df.isna().sum().sum():,}")
    print(f"Duplicate rows    : {df.duplicated().sum():,}")
    
    print("\nColumn names:")
    print(list(df.columns))
```

Cell 4 — Validate AISHE

```python
# AISHE Excel validation

aishe_files = list(AISHE_DIR.glob("*.xlsx"))

print("AISHE DATASET CHECK")
print("=" * 60)

if len(aishe_files) == 0:
    print("❌ No AISHE Excel file found.")

else:
    for file_path in aishe_files:
        print(f"\nFile: {file_path.name}")
        
        excel_file = pd.ExcelFile(file_path)
        
        print(f"Number of sheets: {len(excel_file.sheet_names)}")
        
        print("\nSheet names:")
        for sheet in excel_file.sheet_names:
            print(f"  ✅ {sheet}")
```

Cell 5 — Inspect Relevant AISHE Sheets

```python
# Inspect only the AISHE sheets relevant to our project

relevant_sheets = [
    "2University-Specialisation",
    "3College",
    "6TotalEnr",
    "11Programme",
    "12UGDisc",
    "13PGDisc"
]

print("AISHE RELEVANT SHEETS")
print("=" * 70)

excel_file = pd.ExcelFile(aishe_files[0])

for sheet in relevant_sheets:
    
    if sheet in excel_file.sheet_names:
        
        df = pd.read_excel(
            aishe_files[0],
            sheet_name=sheet,
            nrows=5
        )
        
        print(f"\n{'=' * 70}")
        print(f"SHEET: {sheet}")
        print(f"{'=' * 70}")
        
        print(f"Columns: {len(df.columns)}")
        print("Column names:")
        print(list(df.columns))
        
        print("\nFirst 5 rows:")
        display(df.head())
        
    else:
        print(f"\n❌ Sheet not found: {sheet}")
```

Task 5 — Final AISHE validation cell

```python
# Find actual AISHE sheet names related to colleges, UG and PG

keywords = ["college", "ug", "under", "pg", "post"]

print("RELEVANT AISHE SHEETS")
print("=" * 70)

for sheet in excel_file.sheet_names:
    sheet_lower = sheet.lower()

    if any(keyword in sheet_lower for keyword in keywords):
        print("✅", sheet)
```

```python
# ============================================================
# INSPECT JOB MARKET DATASET
# ============================================================

import pandas as pd

# Use the path Python found
POSTINGS_FILE = PROJECT_DIR / "data" / "raw" / "job_market" / "postings.csv"

# Read only a small sample
job_sample = pd.read_csv(
    POSTINGS_FILE,
    nrows=1000,
    low_memory=False
)

print("JOB MARKET SAMPLE")
print("=" * 70)

print("Rows loaded:", len(job_sample))
print("Columns:", len(job_sample.columns))

print("\nCOLUMN NAMES:")
for col in job_sample.columns:
    print("•", col)

print("\nFIRST 5 ROWS:")
display(job_sample.head())
```

```python
# ============================================================
# JOB MARKET — IMPORTANT COLUMN QUALITY CHECK
# ============================================================

important_cols = [
    "job_id",
    "company_name",
    "title",
    "description",
    "location",
    "skills_desc",
    "formatted_experience_level",
    "work_type",
    "remote_allowed",
    "min_salary",
    "max_salary",
    "normalized_salary",
    "currency",
    "listed_time",
    "job_posting_url"
]

quality = pd.DataFrame({
    "column": important_cols,
    "missing_count": [job_sample[col].isna().sum() for col in important_cols],
    "missing_percent": [
        round(job_sample[col].isna().mean() * 100, 2)
        for col in important_cols
    ],
    "data_type": [str(job_sample[col].dtype) for col in important_cols]
})

display(quality)
```

```python
# ============================================================
# FINAL JOB MARKET VALIDATION
# ============================================================

print("JOB MARKET FINAL CHECK")
print("=" * 70)

print("Total rows:", len(job_sample))
print("Total columns:", len(job_sample.columns))

print("\nRequired columns:")
required_cols = [
    "job_id",
    "title",
    "description",
    "location",
    "work_type",
    "job_posting_url"
]

for col in required_cols:
    print(f"{'✅' if col in job_sample.columns else '❌'} {col}")

print("\nDuplicate job IDs:", job_sample["job_id"].duplicated().sum())

print("\nUnique job titles:", job_sample["title"].nunique())

print("\nTop 15 job titles:")
print(job_sample["title"].value_counts().head(15))
```
